<h3 style="text-align: center; background: #3f51b5; color: white; padding: 18px; border-radius: 8px; margin: 25px 0; font-family: 'Segoe UI', sans-serif; font-weight: 600; font-size: 1.4em;">
Installing Basic Libraries
</h3>

```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('datasets\\Titanic.csv') 
df
```

```python
# Drop irrelevant columns
df = df[["Survived", "Pclass", "Sex", "Age","Embarked"]]

# Drop missing values
df = df.dropna()

# Encode categorical variable
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"C":0, "Q":1, "S":2})
```

```python
df
```

<h3 style="text-align: center; background: #3f51b5; color: white; padding: 18px; border-radius: 8px; margin: 25px 0; font-family: 'Segoe UI', sans-serif; font-weight: 600; font-size: 1.4em;">
Creating the Random Forest Model
</h3>

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[["Pclass", "Sex", "Age", "Embarked"]]
y = df["Survived"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)
```

<h3 style="text-align: center; background: #3f51b5; color: white; padding: 18px; border-radius: 8px; margin: 25px 0; font-family: 'Segoe UI', sans-serif; font-weight: 600; font-size: 1.4em;">
Accuracy Test
</h3>

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h3>Accuracy Result</h3>
<hr>
<h4> When the result of the accuracy is above 0.6 it means the model is picking up patterns it is now <strong> LEARNING </strong></h4>

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

</h3>

```python
import matplotlib.pyplot as plt

importances = model.feature_importances_
features = X_train.columns

plt.barh(features, importances)
plt.xlabel("Feature Importance")
plt.title("Random Forest Feature Importance")
plt.show()
```

<h3 style="text-align: center; background: #3f51b5; color: white; padding: 18px; border-radius: 8px; margin: 25px 0; font-family: 'Segoe UI', sans-serif; font-weight: 600; font-size: 1.4em;">
Titanic Survivability Using Logistic Regression </h3>

```python
# Keep only useful columns

df2 = pd.read_csv("datasets\\Titanic.csv")
df2
```

```python
# Keep only useful columns
df2 = df2[["Survived", "Pclass", "Sex", "Age", "Fare", "Embarked"]]

# Drop missing values
df2 = df2.dropna()
```

```python
df2
```

```python
# Encode 'Sex' and 'Embarked'
df2["Sex"] = df2["Sex"].map({"male": 0, "female": 1})
df2["Embarked"] = df2["Embarked"].map({"S": 0, "C": 1, "Q": 2})
```

```python
df2
```

<h3 style="text-align: center; background: #3f51b5; color: white; padding: 18px; border-radius: 8px; margin: 25px 0; font-family: 'Segoe UI', sans-serif; font-weight: 600; font-size: 1.4em;">
Selecting X and y </h3>

```python
X = df2[["Pclass", "Sex", "Age", "Fare", "Embarked"]]
y = df2["Survived"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

Manual Testing </h3>
